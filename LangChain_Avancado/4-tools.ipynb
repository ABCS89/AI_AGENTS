## Calcular distancia com KM estatica

```python
from langchain.agents import tool
from langchain.pydantic_v1 import BaseModel, Field

```

```python
class CalcularDistanciaArgs(BaseModel):
    cidade_origem: str = Field(description="Cidade de origem",
                               examples=["São Paulo", "Rio de Janeiro"])
    
    cidade_destino: str = Field(description="Cidade de destino",
                               examples=["Belo Horizonte", "Curitiba"])
    
@tool(args_schema=CalcularDistanciaArgs)

def calcular_distancia(cidade_origem: str, cidade_destino: str):
    """
    Calcula a distância entre duas cidades.
    """
    return f"A distância entre {cidade_origem} e {cidade_destino} é de 500 km."
```

```python
calcular_distancia.invoke({
    "cidade_origem": "São Paulo",
    "cidade_destino": "Belo Horizonte"
})
```

## Calculadora

```python
from langchain.agents import tool
from langchain.pydantic_v1 import BaseModel, Field

class OperacaoMatematicaArgs(BaseModel):
    numero1: float = Field(description="Primeiro número",
                           examples=[10, 20.5])
    
    numero2: float = Field(description="Segundo número",
                           examples=[5, 3.5])
    
    operacao: str = Field(description="Operação matemática a ser realizada",
                          examples=["soma", "subtracao", "multiplicacao", "divisao"])

@tool(args_schema=OperacaoMatematicaArgs)
def realizar_calculo(numero1: float, numero2: float, operacao: str):
    """
    Realiza uma operação matemática entre dois números.
    """
    if operacao == "soma":
        return numero1 + numero2
    elif operacao == "subtracao":
        return numero1 - numero2
    elif operacao == "multiplicacao":
        return numero1 * numero2
    elif operacao == "divisao":
        if numero2 == 0:
            return "Erro: Divisão por zero não é permitida."
        return numero1 / numero2
    else:
        return "Operação inválida. Use 'soma', 'subtracao', 'multiplicacao' ou 'divisao'."
    
realizar_calculo
```

```python
realizar_calculo.invoke({
    "numero1": 10,
    "numero2": 0,
    "operacao": "divisao"
})
```
