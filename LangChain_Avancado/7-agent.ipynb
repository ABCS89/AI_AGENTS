```python
from langchain.agents import initialize_agent, Tool, AgentType
from langchain.llms import OpenAI

llm = OpenAI(temperature=0)

def resposta_simples(query):
    return f"A resposta para a '{query}' é Paris."

tools = [
    Tool(
        name="respostas simples",
        func=resposta_simples,
        description="Use esta ferramenta para responder perguntas simples sobre a capital da França."
    )
]
```

```python
agent = initialize_agent(tools, llm, agent_type=AgentType.ZERO_SHOT_REACT_DESCRIPTION, verbose=True)

input_query = "Qual é a capital da França?"
response = agent.run(input_query)
print(response)
```

```python
from langchain.agents import AgentExecutor
import sqlite3
```

```python
def create_database():
    conn = sqlite3.connect(':memory:')
    c = conn.cursor()
    c.execute('''CREATE TABLE users (id INTEGER PRIMARY KEY, name TEXT, age INTERGER)''')
    c.execute("INSERT INTO users (name, age) VALUES ('Alice', 30), ('Bob', 25)")
    conn.commit()
    return conn

def search_user_db(user_name):
    conn = create_database()
    c = conn.cursor()
    c.execute("SELECT name, age FROM users WHERE name = ?", (user_name, ))
    result = c.fetchone()
    conn.close()
    if result:
        return f"O usuário {result[0]} tem a Idade de {result[1]} anos"
    else:
        return "Usuário não encontrado"
    
def sum_numbers(numbers):
    if isinstance(numbers, str):
        numbers = eval(numbers)
    
    if isinstance(numbers, list):
        total = sum(numbers)
        return f"A soma dos números {numbers} é {total}"
    else:
        return "Entrada inválida. Por favor, forneça uma lista de números."
```

```python
tools = [
    Tool(
        name="Buscar Usuário no DB",
        func=search_user_db,
        description="Use esta ferramenta para buscar informações de usuários em um banco de dados SQLite."
    ),
    
    Tool(
        name="Somar Números",
        func=sum_numbers,
        description="Use esta ferramenta para somar uma lista de números."
    )
]
```

```python
agent = initialize_agent(tools, llm, agent_type=AgentType.ZERO_SHOT_REACT_DESCRIPTION, verbose=True)

input_query = "qual é a soma dos numero 5, 10, 20 e 30? e qual a idade de Alice?"
response = agent.run(input_query)
print(response)
```
