```python
import pandas as pd
from langchain_experimental.agents.agent_toolkits import create_pandas_dataframe_agent
from langchain_openai import ChatOpenAI
```

```python
df_temp = pd.read_csv("http://datahub.io/core/global-temp/r/0.csv")
df_temp.head()
```

```python
chat = ChatOpenAI(model="gpt-3.5-turbo-0125")
agent_temp = create_pandas_dataframe_agent(
    chat,
    df_temp,
    verbose=True,
    agent_type="tool-calling",
    allow_dangerous_code=True
)
```

```python
# 1- Qual a temperatura média global do último ano?
resposta = agent_temp.invoke({"input": "Qual a temperatura média global do último ano?"})
print(resposta["output"])
```

```python
# 2- Qual foi o ano com a maior temperatura média?
resposta = agent_temp.invoke({"input": "Qual foi o ano com a maior temperatura média?"})
print(resposta["output"])
```

```python
# 3- qual a variação de temperatura, da menor para a maior, a partir do inicio das medições registradas?
resposta = agent_temp.invoke({"input": "qual a variação de temperatura, da menor para a maior, a partir do inicio das medições registradas?"})
print(resposta["output"])
```

```python
from langchain_community.utilities.sql_database import SQLDatabase
from langchain_community.agent_toolkits.sql.base import create_sql_agent
from langchain_openai import ChatOpenAI
```

```python
db_sakila = SQLDatabase.from_uri("sqlite:///files/sakila_master.db")
chat = ChatOpenAI(model="gpt-3.5-turbo-0125")
agent_sakila = create_sql_agent(
    chat,
    db=db_sakila,
    verbose=True,
    agent_type="tool-calling"
)
```

```python
response = agent_sakila.invoke({"input": "quais os filmes mais alugados?"})
print(response)
```
