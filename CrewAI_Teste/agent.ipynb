```python
from dotenv import load_dotenv
_ = load_dotenv()

from crewai import Crew, Process, Agent, Task
```

```python
planejador_de_viagem = Agent(
    role="Planejador de Viagem",
    goal="Planejar todos os detahles de uma viagem, incluindo roteiros e atividades.",
    backstory="""
    Você é um especialista em planejamento de viagens, sempre em busca de novas aventuiras e experiencias. Seu obbjetivco é garantir que os detalhers da viagemsejam organizados de maneira eficiente e agradavel.
    """,
    
    verbose=True   
)
```

```python
orcamentista = Agent(
    role="Orçamentista",
    goal="Criar um orçamento detalhado para a viagem planejada, considerando transporte, hospedagem, alimentação e atividades.",
    backstory="""
    Você é um especialista em finanças de viagens, com um talento especial para encontrar as melhores
    ofertas e garantir que cada centavo seja bem gasto. Seu objetivo é criar um orçamento que maximize a experiência de viagem sem comprometer a qualidade,
    criando estimativas precisas e detalhadas.
    """,
)
```

```python
planeja_roteiro = Task(
    name="Planejar Roteiro de Viagem",
    description="Desenvolver um roteiro detalhado para a viagem, incluindo destinos, atividades diárias e pontos turísticos.",
    agent=planejador_de_viagem,
    expected_output="""
    Um roteiro detalhado que inclui:
    - Destinos principais
    - Atividades diárias
    - Pontos turísticos
    - Duração de cada atividade
    - Sugestões de restaurantes e locais para visitar
    """
)
```

```python
estima_orcamento = Task(
    name="Estimar Orçamento de Viagem",
    description="Criar um orçamento detalhado para a viagem planejada, considerando transporte, hospedagem, alimentação e atividades.",
    agent=orcamentista,
    expected_output="""
    Um orçamento detalhado que inclui:
    - Custos de transporte (voos, aluguel de carro, transporte público)
    - Custos de hospedagem (hotéis, aluguéis de temporada)
    - Custos de alimentação (restaurantes, compras de supermercado)
    - Custos de atividades (entradas, passeios, eventos)
    """
)
```
